```python
import pandas as pd
from sqlalchemy import create_engine

# Add \Sample_Superstore.csv at the end
file_path = r"C:\Users\Aarya\OneDrive\Desktop\Superstore Sales & Profit Analysis\Dataset\Sample_Superstore.csv"

# 1. Read CSV
df = pd.read_csv(file_path, encoding='windows-1252')
print(f"Total rows read from CSV: {len(df)}")

# 2. Connect to MySQL with your password
engine = create_engine('mysql+pymysql://root:root@localhost/superstore_db')

# 3. Write directly into superstore_full
df.to_sql('superstore_full', con=engine, if_exists='replace', index=False)

print("Import to superstore_full complete!")
```
